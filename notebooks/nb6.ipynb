Splitsen van een lijst in twee lijsten die hooguit 1 in lengte verschillen.

````haskell
splits :: [Int] -> ([Int], [Int]) -> ([Int], [Int]) 
splits [] a = a
splits (x:xs) (lsta, lstb) = splits xs (lstb, x:lsta)
````

````haskell
splits [1..10] ([],[])
````

````haskell
splits [] ([], [])
````

Als eerste voorbeeld van een functie definiëren we de functie `sqr`, voor het kwadrateren van een getal. Deze definitie ziet er vrijwel uit als de gebruikelijke wiskundige definitie:

$$
f(x) = x * x
$$

````haskell
sqr :: Int -> Int
sqr x = x * x
````

We kunnen deze functie *toepassen* op het *argument* 3. Dit is een voorbeeld van functie-applicatie ofwel functie-toepassing.

````haskell
sqr 3
````

De betekenis van deze toepassing (applicatie) van de functie `sqr` kunnen we omschrijven met de volgende uitdrukking:

````haskell
let x = 3 in x * x
````

Dit kunnen we als volgt uitrekenen:

```
    let x = 3 in x * x
==  -- vervang x door de definitie
    3 * 3
==  -- rekenen
    9
```
